## 2. Target or proxy

***What would you predict? Where does that label come from — observed outcome or a defined rule?***

For the scoring model, the goal is to predict the % decrease in impressions and CTR for a page between two ten-week periods: pre-cutoff and post-cutoff. We will use historical data to train the model because we already know the outcome, thus the model can learn the patterns through observation rather than a human rule.

For the multi-label classification model, the goal is to predict for each fix-type, whether a fix was applied at a cutoff AND was followed by a % increase in the relevant metric(s) over the 10 weeks after vs. before (five separate labels). The label comes from a human-applied fix combined with a paired increase in metrics relevant to that fix type. It's observed rather than rule-defined because we observe the effects of a human-applied fix in the data to determine if a fix worked.

## 3. Success metric

***One metric you can defend. What number means 'good'?***

For the scoring model, we can use precision@K to determine if the model has improved relative to past iterations of the model or the baseline rule-based algorithm. A good goal would be a value of 0.8 for precision@K — a prior decision tree model on starter data reached 0.8–0.9 precision@K (at K=20/50), and we expect the added features and complexity of the full model to maintain that level of performance rather than degrade it, giving us confidence the model can hold up as complexity increases.

For the classification model, we can use per-label AUC to check how well the model ranks pages that need a fix above pages that don't. A good goal would be a value of 0.8 for AUC, meaning the model correctly ranks a page that needs a fix above one that doesn't about 4 out of 5 times.
